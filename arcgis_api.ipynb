# 1. Comparing ArcGIS API for Python Classification Models

```python
# Import the packages
import pandas as pd
from arcgis.learn import prepare_textdata
from arcgis.learn.text import TextClassifier
```

```python
# Change the encoding to UTF-8
df = pd.read_csv('./data/top_1000_tags.csv', encoding='latin')
df.to_csv('./data/top_1000_tags.csv', encoding='utf-8')
```

```python
DATA_ROOT = "./data/"
```

## 1.1 Character Perspective Category Annotation

```python
char_data = prepare_textdata(
    DATA_ROOT,
    "classification",
    train_file="top_1000_tags.csv",
    text_columns="Tag",
    label_columns="Character",
    batch_size=32
)
```

```python
print(char_data.classes) # Print the labels
```

```python
char_data.show_batch(rows=10)
```

```python
print(TextClassifier.supported_backbones) # Print the available models
```

```python
print(TextClassifier.available_backbone_models("BERT")) # Print the available BERT models
```

```python
char_model = TextClassifier(char_data, backbone="bert-base-uncased") # Pick the BERT base uncased model
```

```python
char_model.fit(epochs=2) # Fit the model
```

```python
char_model.accuracy() # Show the accuracy score
```

```python
char_model.metrics_per_label() # Show the precision, recall, and F1 scores and the support for each label
```

```python
char_model.show_results(32) # Show the prediction results
```

```python
# Test the model
char_test_tag = "AvaLance"
print(char_model.predict(char_test_tag))
```
