# 1. Create the Annotated Additional Tags DataFrame

```python
# Import package
import polars as pl
```

```python
# Set the string length to 10000
pl.Config.set_fmt_str_lengths(10000)
```

```python
df_additional_tags = pl.read_csv("./data/df_additional_tags.csv")
df_additional_tags
```

```python
df_char = pl.read_csv("./data/df_character.csv")
df_nar = pl.read_csv("./data/df_narrative.csv")
df_content_control = pl.read_csv("./data/df_content_control.csv")

print(df_char)
print(df_nar)
print(df_content_control)
```

```python
# Rename the Tag column
df_content_control = df_content_control.rename({"Tag": "Additional Tags"})
df_content_control
```

```python
# Concatenate the DataFrames
df_annotated_tags = pl.concat([df_char, df_nar, df_content_control], how="align")
df_annotated_tags
```

```python
# Remove the Confidence column
df_annotated_tags = df_annotated_tags.drop(pl.col("Confidence"))
df_annotated_tags
```

```python
# Concatenate the DataFrames
df_annotated_additional_tags = pl.concat([df_additional_tags, df_annotated_tags], how="align")
df_annotated_additional_tags
```

```python
# Save the df_annotated_tags DataFrame as a CSV file
df_annotated_additional_tags.write_csv('data/df_annotated_additional_tags.csv')
```
